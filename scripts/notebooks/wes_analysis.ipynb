```python
from datasets.Harmonized_Datasets import harmonize
import pandas as pd
import numpy as np
import statsmodels.api as sm
```

```python
mf_ad = harmonize()['Moffitt']
ml_results_df = pd.read_csv("../../outputs/predictions_for_wes_analysis.csv").set_index('samples')
maf = pd.read_csv("../../data/Moffitt/Whole_Exome/MAF.maf", sep='\t').set_index('sampleID')
```

```python
maf=maf[maf['Hugo_Symbol'] !='Unknown']
```

```python
non_synonymous_mutations = ['Frame_Shift_Ins', 'Frame_Shift_Del', 'Nonsense_Mutation', 'In_Frame_Ins', 'Nonstop_Mutation', 'In_Frame_Del', 'Missense_Mutation']
```

```python
maf
```

```python
maf = maf[maf['Variant_Classification'].isin(non_synonymous_mutations)]
```

```python
#ras_genes = ['HRAS', "KRAS", "NRAS", "BRAF"]
ras_genes = ["KRAS"]
maf = maf[maf['Hugo_Symbol'].isin(ras_genes)]
```

```python
maf
```

```python
mf_ad.obs
```

```python
tmb = {}
for wes_id, rna_id in zip(mf_ad.obs['WES_SampleID'], mf_ad.obs.index):
    if not pd.isna(wes_id):
        if wes_id in maf.index:
            tmb[rna_id] = [len(maf.loc[wes_id]), wes_id]
        
tmb = pd.DataFrame(tmb, index=['TMB', 'WES_SampleID']).T
```

```python
tmb
```

```python
mf_ad = mf_ad[mf_ad.obs.index.isin(tmb.index)] # this will also align the indices
```

```python
mf_ad.obs['TMB'] = tmb['TMB']
```

```python
mf_ad.obs[['y_true', 'y_pred', 'y_prob']] = ml_results_df[['y_true', 'y_pred', 'y_prob']]
```

```python
# remove 'moderate response' samples but I can try using the model ot predict their response
mf_ad = mf_ad[mf_ad.obs['y_pred'].notna()]
```

```python
mf_ad.obs
```

```python
mf_ad.obs['t11_14'] = pd.Categorical(mf_ad.obs['t11_14'])
```

```python
# Regression Analysis
mf_ad_og = mf_ad.copy()
mf_ad = mf_ad[mf_ad.obs['t11_14'].notna()]

# single coefficients
y = mf_ad.obs['ABT-199']



model_1 = sm.OLS(y, sm.add_constant(mf_ad.obs['t11_14'])).fit(cov_type='HC3')
model_2 = sm.OLS(y, sm.add_constant(mf_ad.obs['y_prob'])).fit(cov_type='HC3')
model_3 = sm.OLS(y ,sm.add_constant(mf_ad.obs['TMB'])).fit(cov_type='HC3')

# two coefficients
x_4 = sm.add_constant(np.column_stack(
   [ mf_ad.obs['t11_14'], mf_ad.obs['y_prob'] ]
        )  
    )
model_4 = sm.OLS(y,x_4).fit(cov_type='HC3')

x_5 = sm.add_constant(np.column_stack(
   [ mf_ad.obs['t11_14'], mf_ad.obs['TMB'] ]
        )  
    )
model_5 = sm.OLS(y,x_5).fit(cov_type='HC3')

x_6 = sm.add_constant(np.column_stack([mf_ad.obs['y_prob'], mf_ad.obs['TMB']]))
model_6 = sm.OLS(y, x_6).fit(cov_type='HC3')

# three coefficients 
x_7 = sm.add_constant(np.column_stack(
   [ mf_ad.obs['t11_14'], mf_ad.obs['TMB'], mf_ad.obs['y_prob'] ]
        )  
    )
model_7 = sm.OLS(y,x_7).fit(cov_type='HC3')

print(f't11;14 status only: { model_1.rsquared}')

print(f'model probs only: {model_2.rsquared}')


print(f'tumor burden only: {model_3.rsquared}')

print(f'model probs and t11;14 status: {model_4.rsquared}')

print(f't11;14 status + tumor burden: {model_5.rsquared}')

print(f'model probability + tumor burden: {model_6.rsquared}')

print(f'model probability + tumor burden + t11;14 status: {model_7.rsquared}')




```

```python
model_4.wald_test('x2 = 0', use_f=True)
```

```python

# does y_prob + TMB add beyond t11;14?
print(model_7.wald_test('x2 = 0, x3 = 0', use_f=True))

# does  TMB add beyond t11;14?
print(model_7.wald_test('x2 = 0', use_f=True))

# does y_prob add beyond t11;14 + TMB? 
print(model_7.wald_test('x3 = 0', use_f=True))
```

```python
mutation_matrix = {}

for gene in maf['Hugo_Symbol']:
    gene_maf = maf[maf['Hugo_Symbol'] == gene]
    for variant in gene_maf['Variant_Classification']:
        col_name = str(gene)+str(variant)
        gene_variant_maf = gene_maf[gene_maf['Variant_Classification'] == variant]
        mutation_matrix[col_name] = []
        for wes_id, rna_id in zip(mf_ad.obs['WES_SampleID'], mf_ad.obs.index):
            if not pd.isna(wes_id):
                if wes_id in gene_variant_maf.index:
                    count = [len(gene_variant_maf.loc[wes_id])]
                    mutation_matrix[col_name].append(count)
```
